```python
def eliminar_duplicados_exactos(df):
    """
    Elimina solo duplicados EXACTOS manteniendo la primera ocurrencia
    """
    print("🔍 Analizando duplicados exactos...")
    # Contar filas antes
    filas_antes = len(df)
    
    # Identificar duplicados exactos en todas las columnas
    duplicados_exactos = df.duplicated(keep='first')
    
    # Mostrar información antes de eliminar
    print(f"📊 Filas antes: {filas_antes}")
    print(f"🔴 Duplicados exactos encontrados: {duplicados_exactos.sum()}")
    
    # Eliminar duplicados exactos (mantener solo la primera ocurrencia)
    df_limpio = df.drop_duplicates(keep='first')
    
    # Contar filas después
    filas_despues = len(df_limpio)
    eliminadas = filas_antes - filas_despues
    
    print(f"✅ Filas después: {filas_despues}")
    print(f"🗑️ Filas eliminadas: {eliminadas}")
    
    return df_limpio

# Aplicar la función al dataset work_orders
work_orders_sin_duplicados = eliminar_duplicados_exactos(work_orders)

# Actualizar la variable work_orders con los datos sin duplicados
work_orders = work_orders_sin_duplicados

# Guardar cambio en el dataset limpio
work_orders.to_csv('work_orders_clean.csv', index=False)
print("\n💾 cambio guardado en 'work_orders_clean.csv'")
```

```python
# Imputar país faltante usando la distribución proporcional existente
paises_distribucion = work_orders['pais'].value_counts(normalize=True)
print("Distribución actual de países:")
print(paises_distribucion)

# Imputar valores faltantes manteniendo la proporción
np.random.seed(42) 

# Obtener índices de valores faltantes
indices_faltantes = work_orders['pais'].isnull()
n_faltantes = indices_faltantes.sum()

# Generar países aleatorios según la distribución existente
paises_imputados = np.random.choice(
    paises_distribucion.index,
    size=n_faltantes,
    p=paises_distribucion.values
)

# Aplicar la imputación 
work_orders.loc[indices_faltantes, 'pais'] = paises_imputados
print(f"✅ {n_faltantes} valores de país imputados")

# Guardar cambio en el dataset limpio
work_orders.to_csv('work_orders_clean.csv', index=False)
print("\n💾 cambio guardado en 'work_orders_clean.csv'")
```

```python
#Imputar valores faltantes en precio_unitario 
def imputar_precios_con_datos_reales(df):
    """
    Imputa precio_unitario usando datos reales del cliente:
    precio_unitario = (monto_total + monto_descuento) / cantidad
    """
    print("🎯 IMPLEMENTANDO ESTRATEGIA DE IMPUTACIÓN INTELIGENTE")
    print("="*60)
    
    # Identificar casos con precio_unitario faltante
    mask_precio_faltante = df['precio_unitario'].isnull()
    casos_a_imputar = df[mask_precio_faltante].copy()
    
    print(f"📊 Casos a imputar: {len(casos_a_imputar)}")
    
    # Calcular precio_unitario usando datos reales
    precio_calculado = (casos_a_imputar['monto_total'] + casos_a_imputar['monto_descuento']) / casos_a_imputar['cantidad']
    
    # Aplicar la imputación
    df.loc[mask_precio_faltante, 'precio_unitario'] = precio_calculado
    
    # Mostrar resultados
    print(f"✅ {len(casos_a_imputar)} precios imputados con datos reales")
    print(f"📊 Precios faltantes restantes: {df['precio_unitario'].isnull().sum()}")
    
    return df

# Ejecutar estrategia
work_orders_imputado_real = imputar_precios_con_datos_reales(work_orders.copy())
work_orders = imputar_precios_con_datos_reales(work_orders)

# Guardar cambios en dataset clean
work_orders.to_csv('work_orders_clean.csv', index=False)
print("\n💾 Dataset guardado en 'work_orders_clean.csv'")
```

```python
# DETECCIÓN DE OUTLIERS USANDO MÉTODO IQR
def detectar_outliers_precios(df, columna='precio_unitario'):
    """
    Detecta outliers usando el método del Rango Intercuartílico (IQR)
    """
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    
    # Límites para outliers
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    
    # Identificar outliers
    outliers_mask = (df[columna] < limite_inferior) | (df[columna] > limite_superior)
    outliers = df[outliers_mask]
    
    print(f"📊 RESULTADOS DETECCIÓN DE OUTLIERS:")
    print(f"   • Q1 (25%): ${Q1:.2f}")
    print(f"   • Q3 (75%): ${Q3:.2f}")
    print(f"   • IQR: ${IQR:.2f}")
    print(f"   • Límite inferior: ${limite_inferior:.2f}")
    print(f"   • Límite superior: ${limite_superior:.2f}")
    print(f"   • Outliers detectados: {len(outliers)}")
    
    return outliers, limite_inferior, limite_superior

# Ejecutar detección
outliers_detectados, limite_inf, limite_sup = detectar_outliers_precios(work_orders)
```

```python
# PARTE 1: Totales del funnel
# ======================
# Eventos disponibles
query_eventos_disponibles = '''
SELECT 
    nombre_evento,
    COUNT(DISTINCT id_usuario) as usuarios_unicos
FROM events
GROUP BY nombre_evento
ORDER BY usuarios_unicos DESC;
'''
eventos_disponibles = pd.read_sql(query_eventos_disponibles, con=engine)
eventos_disponibles
```

```python
# PARTE 2: Conversiones
# ======================

query_funnel_completo = '''
WITH funnel_analysis AS (
    SELECT
        'first_visit' as etapa,
        1 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'first_visit'
    
    UNION ALL
    
    SELECT
        'select_item' as etapa,
        2 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'select_item'
    
    UNION ALL
    
    SELECT
        'add_to_cart' as etapa,
        3 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'add_to_cart'
    
    UNION ALL
    
    SELECT
        'begin_checkout' as etapa,
        4 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'begin_checkout'
    
    UNION ALL
    
    SELECT
        'add_payment_info' as etapa,
        5 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'add_payment_info'
    
    UNION ALL
    
    SELECT
        'purchase' as etapa,
        6 as orden_etapa,
        COUNT(DISTINCT id_usuario) as usuarios_unicos
    FROM events
    WHERE nombre_evento = 'purchase'
),
funnel_with_conversion AS (
    SELECT
        etapa,
        orden_etapa,
        usuarios_unicos,
        LAG(usuarios_unicos) OVER (ORDER BY orden_etapa) as usuarios_etapa_anterior,
        CASE 
            WHEN LAG(usuarios_unicos) OVER (ORDER BY orden_etapa) IS NULL THEN 100.0
            ELSE ROUND(
                (usuarios_unicos * 100.0) / LAG(usuarios_unicos) OVER (ORDER BY orden_etapa), 
                2
            )
        END as tasa_conversion,
        CASE 
            WHEN LAG(usuarios_unicos) OVER (ORDER BY orden_etapa) IS NULL THEN 0
            ELSE LAG(usuarios_unicos) OVER (ORDER BY orden_etapa) - usuarios_unicos
        END as usuarios_perdidos,
        ROUND(
            (usuarios_unicos * 100.0) / FIRST_VALUE(usuarios_unicos) OVER (ORDER BY orden_etapa), 
            2
        ) as conversion_desde_inicio
    FROM funnel_analysis
)
SELECT
    etapa,
    usuarios_unicos,
    usuarios_etapa_anterior,
    tasa_conversion,
    usuarios_perdidos,
    conversion_desde_inicio
FROM funnel_with_conversion
ORDER BY orden_etapa;
'''
# Ejecutar la consulta usando pandas
funnel_completo = pd.read_sql(query_funnel_completo, con=engine)

# Crear formato personalizado - VERSIÓN CORREGIDA
print("ANÁLISIS DE CONVERSIÓN DEL FUNNEL")
print("=" * 60)
print(f"{'ETAPA':<18} {'USUARIOS':<10} {'CONVERSIÓN':<12} {'PERDIDOS':<10}")
print("-" * 60)

for _, row in funnel_completo.iterrows():
    etapa = str(row['etapa'])  # ✅ Convertir a string
    usuarios = f"{int(row['usuarios_unicos']):,}"
    conversion = f"{float(row['tasa_conversion']):.2f}%"
    perdidos = f"{int(row['usuarios_perdidos']):,}"
    
    print(f"{etapa:<18} {usuarios:<10} {conversion:<12} {perdidos:<10}")
```

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---

1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Verificar si hay usuarios duplicados
query_duplicados = '''
SELECT id_usuario,
       count(fecha_registro) AS registros_duplicados
FROM users
GROUP BY id_usuario
HAVING count(fecha_registro) > 1
'''

duplicados = pd.read_sql(query_duplicados, con=engine)
print(f"Usuarios con registros duplicados: {len(duplicados)}")
duplicados.head()
```

```python
# Asignación id_usuario a cohorte
query_cohort_fecha = '''
SELECT id_usuario,
MIN(CAST(fecha_registro AS DATE)) AS cohort_fecha
FROM users
GROUP BY id_usuario
ORDER BY MIN(fecha_registro) ASC;
'''
resultado = pd.read_sql(query_cohort_fecha, con=engine)
resultado
```

```python
# Retención por cohortes
# ======================
query_cohort_retention_final = '''
WITH cohortes AS (
    SELECT
        id_usuario,
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte_mes
    FROM users

),

cohort_size AS (
    SELECT
        cohorte_mes,
        COUNT(DISTINCT id_usuario) AS clientes_iniciales
    FROM cohortes
    GROUP BY cohorte_mes

),

actividad_usuario AS (
    SELECT
        ua.id_usuario,
        c.cohorte_mes,
        MAX(CASE
            WHEN ua.dias_despues_registro >= 1
                 AND ua.activo = 1
            THEN 1 ELSE 0
        END) AS retained_w1,
        MAX(CASE
            WHEN ua.dias_despues_registro >= 8
                 AND ua.activo = 1
            THEN 1 ELSE 0
        END) AS retained_w2,
        MAX(CASE
            WHEN ua.dias_despues_registro >= 15
                 AND ua.activo = 1
            THEN 1 ELSE 0
        END) AS retained_w3
    FROM user_activity ua
    JOIN cohortes c
        ON ua.id_usuario = c.id_usuario
    GROUP BY
        ua.id_usuario,
        c.cohorte_mes
),

retention_summary AS (

    SELECT
        cohorte_mes,
        SUM(retained_w1) AS retained_w1,
        SUM(retained_w2) AS retained_w2,
        SUM(retained_w3) AS retained_w3
    FROM actividad_usuario
    GROUP BY cohorte_mes
)

SELECT
    TO_CHAR(cs.cohorte_mes, 'YYYY-MM') AS cohorte,
    cs.clientes_iniciales,
    rs.retained_w1,
    rs.retained_w2,
    rs.retained_w3,

    ROUND(rs.retained_w1 * 1.0 / cs.clientes_iniciales, 2) AS semana_1,
    ROUND(rs.retained_w2 * 1.0 / cs.clientes_iniciales, 2) AS semana_2,
    ROUND(rs.retained_w3 * 1.0 / cs.clientes_iniciales, 2) AS semana_3

FROM cohort_size cs

LEFT JOIN retention_summary rs
    ON cs.cohorte_mes = rs.cohorte_mes
ORDER BY cs.cohorte_mes;

'''
# Ejecutar la consulta
cohort_retention_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohort_retention_final
```

```python
# Análisis de cohortes
cohort_data = cohort_retention_final.copy()

# Exportar para Power BI
cohort_data.to_csv('cohort_retention_clean.csv', index=False)
```

Hipótesis estadística
   - **H₀ (Hipótesis nula):** La nueva UI NO impacta las conversiones (no hay diferencia)
   - **H₁ (Hipótesis alternativa):** La nueva UI SÍ impacta las conversiones (hay diferencia)
   
**Test estadístico:** Chi-cuadrado: 0.6178 Valor bajo = poca diferencia entre control y tratamiento
Valor p: 0.4319
- Probabilidad de que la diferencia sea casualidad
- 43.19% de probabilidad de que sea solo suerte

Regla de decisión:
- Si p-value < α (0.05) → Rechazamos H₀ (hay diferencia significativa)
- Si p-value ≥ α (0.05) → No rechazamos H₀ (no hay diferencia significativa)
-**Nivel de significancia alpha:** α = 0.05 (95% confianza) - ESTÁNDAR "Con 95% de confianza, NO hay evidencia suficiente de que la nueva UI impacte las conversiones" 
Resultado: p-value (0.4319) > α (0.05)

Conclusión: ❌ No rechazamos H₀
- La nueva UI NO genera un impacto estadísticamente significativo en las conversiones
- La diferencia del 0.6% entre control/tratamiento probablemente es casualidad
